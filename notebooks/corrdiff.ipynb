```python
import zarr
import numpy as np

z = zarr.open(
    "/home/sangonvi/Cefet/repositories/atmoseer/datasets/corrdiff/train.zarr",
    mode="r"
)

y = z["target"]
```

```python
print("global min", np.min(y))
print("global max", np.max(y))
print("global mean", np.mean(y))
```

```python
count5 = 0
count20 = 0

for i in range(len(y)):

    patch = np.expm1(y[i])

    if patch.max() > 5:
        count5 += 1

    if patch.max() > 20:
        count20 += 1

print("patches > 5 dBZ :", count5)
print("patches > 20 dBZ:", count20)
print("total:", len(y))
```

```python
import zarr
import numpy as np

z = zarr.open(
    "/home/sangonvi/Cefet/repositories/atmoseer/datasets/corrdiff/train.zarr",
    mode="r"
)

y = z["target"]

print("global max:", np.max(y))
print("global mean:", np.mean(y))
print(
    "max expm1:",
    np.expm1(np.max(y))
)
```

```python
y_dbz = np.expm1(y)

print("max dBZ:", y_dbz.max())
print("mean dBZ:", y_dbz.mean())

print(
    "fraction rainy:",
    np.mean(y_dbz > 5)
)

print(
    "fraction strong:",
    np.mean(y_dbz > 20)
)
```
