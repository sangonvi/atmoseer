```python
import folium
import leafmap.foliumap as leafmap
import requests
import datetime
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import PIL.Image as Image
import numpy as np
import math
from scipy import stats
```

```python
mapa = leafmap.Map(center= [-22.925778948753702, -43.489029909370046], zoom_start=8, tiles='OpenStreetMap')
colormap = ['magenta', 'red', 'orange', 'yellow']
folium.LatLngPopup().add_to(mapa)
```

```python
lat = [-22.955139]
lon = [-43.248278]

for loc in zip(lat, lon):
    folium.Circle(
        location=loc,
        radius=100,
        fill=True,
        color='black',
        fill_opacity=0.7
    ).add_to(mapa)

for loc in zip(lat, lon):
    folium.Circle(
        location=loc,
        radius=137100,
        fill=True,
        color='gray',
        fill_opacity=0.1
    ).add_to(mapa)
```

```python
import numpy as np

step = 0.25
unique_lats = (-22.19 - np.arange(7) * step)
print(unique_lats)

unique_lons = (-42.35 - np.arange(8) * step)
print(unique_lons)
```

```python

for lat in unique_lats:
    for lon in unique_lons:
        folium.CircleMarker(
            location=[lat, lon],
            radius=2,
            color="blue",
            fill=True,
            fill_color="blue",
            fill_opacity=0.7,
        ).add_to(mapa)

mapa.save("mapa.html")
```
